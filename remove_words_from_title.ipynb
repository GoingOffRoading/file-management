```python
from pathlib import Path
import re


# Edit these settings before running the notebook.
TARGET_DIR = Path(r"Z:\Misc")
TEXT_TO_REMOVE = "Imgur Album "
DRY_RUN = False


def remove_text_from_filenames() -> None:
    if not TARGET_DIR.is_dir():
        raise NotADirectoryError(f"Target directory not found: {TARGET_DIR}")

    pattern = re.compile(re.escape(TEXT_TO_REMOVE), re.IGNORECASE)
    for path in TARGET_DIR.iterdir():
        if not path.is_file() or not pattern.search(path.name):
            continue

        new_name = pattern.sub("", path.name, count=1).strip()
        if not new_name:
            print(f"[SKIP] {path.name}: removing the text would leave an empty name")
            continue

        destination = path.with_name(new_name)
        if destination.exists():
            print(f"[SKIP] {path.name}: destination already exists")
            continue

        if DRY_RUN:
            print(f"[DRY RUN] {path.name} -> {new_name}")
        else:
            path.rename(destination)
            print(f"[RENAMED] {path.name} -> {new_name}")


remove_text_from_filenames()
```

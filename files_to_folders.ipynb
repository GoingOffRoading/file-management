```python
from pathlib import Path
import re
import shutil
```

```python
# Edit these settings before running the script.
SOURCE_DIR = Path(r"Z:\Misc")
EXTENSIONS = {".cbz"}  # Include the dot; matching is case-insensitive.
BATCH_SIZE = 1000  # Number of matching files to process at a time.
DRY_RUN = False

BRACKETED_NAME = re.compile(r"^\[([^\]]+)\]")
```

```python
def move_bracketed_files() -> None:
    if not SOURCE_DIR.is_dir():
        raise NotADirectoryError(f"Source directory not found: {SOURCE_DIR}")
    if BATCH_SIZE < 1:
        raise ValueError("BATCH_SIZE must be at least 1")

    extensions = {extension.lower() for extension in EXTENSIONS}
    files = [
        path
        for path in SOURCE_DIR.iterdir()
        if path.is_file() and path.suffix.lower() in extensions
    ]

    for batch_start in range(0, len(files), BATCH_SIZE):
        batch = files[batch_start:batch_start + BATCH_SIZE]
        batch_number = batch_start // BATCH_SIZE + 1
        print(f"--- Batch {batch_number} ({len(batch)} file(s)) ---")

        for path in batch:
            match = BRACKETED_NAME.match(path.name)
            if match is None:
                print(f"[SKIP] {path.name}: name does not start with [folder name]")
                continue

            folder_name = match.group(1).strip()
            destination_dir = SOURCE_DIR / folder_name
            destination = destination_dir / path.name

            if destination_dir.exists() and not destination_dir.is_dir():
                print(f"[SKIP] {path.name}: {destination_dir} exists and is not a folder")
                continue

            if destination.exists():
                print(f"[SKIP] {path.name}: destination already exists")
                continue

            if DRY_RUN:
                print(f"[DRY RUN] {path.name} -> {destination_dir.name}/")
                continue

            destination_dir.mkdir(exist_ok=True)
            shutil.move(str(path), str(destination))
            print(f"[MOVED] {path.name} -> {destination_dir.name}/")
```

```python
move_bracketed_files()
```
